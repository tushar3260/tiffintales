```python
# Merging in two dataframe

```

```python
data1 = {
    "EmpID": [101, 102, 103, 104, 105],
    "Name": ["Amit", "Priya", "Raj", "Simran", "Karan"],
    "Department": ["HR", "IT", "Finance", "Admin", None]
    
}
df1=pd.DataFrame(data1)
df1
```

```python
data2={
    "EmpID": [101, 102, 103, 106, 107],
    "Salary": [50000, 60000, None, 55000, 45000],
    "Location": ["Delhi", "Noida", "Pune", "Mumbai", "Chennai"]
}
df2=pd.DataFrame(data2)
df2
```

```python
df=pd.merge(df1,df2,on="EmpID",how='outer')
df
```
