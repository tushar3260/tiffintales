```python
def extract(txt):
    check=False
    data=""
    for i in txt:
        if(i==")"):
            check=False
            break
        if check==True:
            data+=i
        if(i=="("):
            check=True
        
    return data
```

```python
df["Episode"]=df["Title"].apply(extract)
df
```

```python
df['Episode']=df['Episode'].str.replace(" eps","")
df
```

```python
def extracttime(txt):
    check=False
    data=""
    for i in range(len(txt)):
        if(txt[i]==")"):
            for j in range(i+1,i+20):
                data+=txt[j]
            return data        
```

```python
df['Time']=df['Title'].apply(extracttime)
df
```
