```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
arr= np.linspace(0,5,11)
x=arr
y=arr**2
```

```python
# Basics of %matplotlib

plt.plot(x,y)
plt.title("graph")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
```

```python
plt.subplot(2,2,1)
plt.plot(x,y)
plt.subplot(2,2,2)
plt.plot(y,x)
plt.subplot(2,2,3)
plt.plot(x**2,y**10)

plt.subplot(2,2,4)
plt.plot(y**2,y**2)


```

```python
arr=plt.figure(figsize=(10,5),dpi=200)
a=arr.add_axes([0.1,0.1,.8,.8])
a.plot(x,y)

b=arr.add_axes([0.5,0.5,.8,.8])
b.plot(x,y)
```
